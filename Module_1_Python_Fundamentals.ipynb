# 2. Control Flow

Control flow statements are used to control the execution of a Python program
based on conditions or repetition.

Topics covered:
- if / elif / else

## 2.1 if / elif / else

Conditional statements are used to execute different blocks of code
depending on whether a condition is true or false.

```python
marks = 78

if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 60:
    grade = "B"
elif marks >= 50:
    grade = "C"
else:
    grade = "Fail"

print("Marks:", marks)
print("Grade:", grade)
```

## 2.2 for Loop

A for loop is used to iterate over a sequence such as a list, tuple or string.

```python
subjects = ["Python", "Machine Learning", "SQL", "Deep Learning"]
completed = ["Python", "SQL"]

for subject in subjects:
    if subject in completed:
        status = "Completed"
    else:
        status = "Pending"
    
    print(subject, ":", status)
```

## 2.3 while Loop

A while loop repeatedly executes a block of code as long as the given condition is true.

```python
count = 1
total = 0

while count <= 5:
    total = total + count
    print("Count:", count, "Total:", total)
    count += 1

print("Final Total:", total)
```

## 2.4 break

The break statement immediately terminates a loop when a specific condition is satisfied.

```python
numbers = [10, 20, 30, 40, 50, 60]

for number in numbers:
    print("Checking:", number)
    
    if number == 40:
        print("Target number found")
        break
```

## 2.5 continue

The continue statement skips the current iteration and moves to the next iteration of the loop.

```python
numbers = range(1, 11)

for number in numbers:
    if number % 2 == 0:
        continue
    
    print("Odd number:", number)
```

## 2.6 pass

The pass statement is a placeholder used when a statement is syntactically required
but no action needs to be performed yet.

```python
numbers = [10, 20, 30, 40]

for number in numbers:
    if number == 30:
        pass
    else:
        print("Processing:", number)

print("Loop completed")
```

## 3.1 List

A list is an ordered and mutable collection.
It allows duplicate values and can store different data types.

```python
skills = ["Python", "SQL", "Machine Learning", "Python"]

print("Original list:", skills)

# Add a new skill
skills.append("Deep Learning")

# Update an existing value
skills[1] = "Advanced SQL"

# Remove a value
skills.remove("Python")

print("Updated list:", skills)
print("Number of skills:", len(skills))
print("First skill:", skills[0])
```

## 3.2 Tuple

A tuple is an ordered and immutable collection.
Once created, its elements cannot be changed.

```python
employee = ("Dileep", "AI/ML Engineer", 25, "Hyderabad")

print("Employee details:", employee)
print("Name:", employee[0])
print("Role:", employee[1])
print("Number of details:", len(employee))

# Tuple unpacking
name, role, age, city = employee

print("\nUnpacked values:")
print("Name:", name)
print("Role:", role)
print("Age:", age)
print("City:", city)
```

## 3.3 Set

A set is an unordered collection of unique elements.
It automatically removes duplicate values.

```python
skills = {"Python", "SQL", "Machine Learning", "Python", "SQL"}

print("Original set:", skills)

skills.add("Deep Learning")
skills.remove("SQL")

print("Updated set:", skills)
print("Number of unique skills:", len(skills))
```

## 3.4 Dictionary

A dictionary stores data in key-value pairs.
It is useful when each value needs a meaningful key.

```python
student = {
    "name": "Dileep",
    "age": 25,
    "course": "AI/ML",
    "marks": 85
}

print("Student details:")
print("Name:", student["name"])
print("Course:", student["course"])
print("Marks:", student["marks"])

# Update marks
student["marks"] = 90

# Add a new key-value pair
student["grade"] = "A"

print("\nUpdated student details:")
for key, value in student.items():
    print(key, ":", value)
```

## 3.5 List Comprehension

List comprehension provides a short and readable way to create a new list
from an existing iterable.

```python
numbers = range(1, 11)

# Create squares of even numbers
even_squares = [number ** 2 for number in numbers if number % 2 == 0]

print("Even numbers:", [number for number in numbers if number % 2 == 0])
print("Squares of even numbers:", even_squares)
```

## 3.6 Dictionary Comprehension

Dictionary comprehension provides a concise way to create dictionaries
using an iterable and an expression.

```python
numbers = range(1, 6)

squares = {number: number ** 2 for number in numbers}

print("Square dictionary:", squares)

# Create a dictionary only for even numbers
even_squares = {
    number: number ** 2
    for number in numbers
    if number % 2 == 0
}

print("Even square dictionary:", even_squares)
```

## 4.1 Function Definition

A function is defined using the `def` keyword.
Functions help us avoid repeating the same code multiple times.

```python
def calculate_average(marks):
    total = sum(marks)
    average = total / len(marks)
    return average


student_marks = [78, 85, 92, 88, 76]

average_marks = calculate_average(student_marks)

print("Marks:", student_marks)
print("Average Marks:", round(average_marks, 2))
```

## 4.2 Arguments

Arguments are values passed to a function when the function is called.

```python
def calculate_salary(basic_salary, bonus):
    total_salary = basic_salary + bonus
    return total_salary


basic = 40000
bonus = 5000

salary = calculate_salary(basic, bonus)

print("Basic Salary:", basic)
print("Bonus:", bonus)
print("Total Salary:", salary)
```

## 4.3 Default Arguments

A default argument has a predefined value.
If the caller does not provide that argument, the default value is used.

```python
def greet_user(name, role="Python Developer"):
    print("Name:", name)
    print("Role:", role)


greet_user("Dileep")

print()

greet_user("Rahul", "Data Scientist")
```

## 4.4 Keyword Arguments

Keyword arguments allow us to pass values using the parameter names.
The order of arguments does not matter when using parameter names.

```python
def employee_details(name, age, department, experience):
    print("Name:", name)
    print("Age:", age)
    print("Department:", department)
    print("Experience:", experience, "years")


employee_details(
    experience=2,
    department="AI/ML",
    name="Dileep",
    age=25
)
```

## 4.5 Variable-Length Arguments - *args

`*args` allows a function to accept any number of positional arguments.
The arguments are received as a tuple.

```python
def calculate_total(*numbers):
    total = 0
    
    for number in numbers:
        total += number
    
    return total


print("Total 1:", calculate_total(10, 20, 30))
print("Total 2:", calculate_total(5, 15, 25, 35, 45))
```

## 4.6 Variable-Length Arguments - **kwargs

`**kwargs` allows a function to accept any number of keyword arguments.
The arguments are received as a dictionary.

```python
def display_profile(**details):
    print("Employee Profile")
    
    for key, value in details.items():
        print(key, ":", value)


display_profile(
    name="Dileep",
    role="AI/ML Engineer",
    location="Hyderabad",
    experience=2
)
```

## 4.7 Lambda Functions

A lambda function is a small anonymous function.
It can take multiple arguments but contains a single expression.

```python
numbers = [2, 4, 6, 8, 10]

square = lambda x: x ** 2

squared_numbers = [square(number) for number in numbers]

print("Original numbers:", numbers)
print("Squared numbers:", squared_numbers)
```

## 4.8 Recursive Functions

A recursive function is a function that calls itself.
It must have a base condition to stop the recursion.

```python
def factorial(number):
    if number == 0 or number == 1:
        return 1
    
    return number * factorial(number - 1)


number = 5
result = factorial(number)

print("Number:", number)
print("Factorial:", result)
```

## 5.1 Iterators

An iterator is an object that allows us to access elements one at a time.

The `iter()` function creates an iterator, and `next()` retrieves
the next available element.

```python
numbers = [10, 20, 30, 40, 50]

number_iterator = iter(numbers)

print("First value:", next(number_iterator))
print("Second value:", next(number_iterator))
print("Third value:", next(number_iterator))

print("\nRemaining values:")
for number in number_iterator:
    print(number)
```

## 5.2 Generators

A generator is a special type of iterator that produces values one at a time.
Generators are memory efficient because they do not store all values at once.

```python
def square_generator(numbers):
    for number in numbers:
        yield number ** 2


numbers = [1, 2, 3, 4, 5]

squares = square_generator(numbers)

for square in squares:
    print("Square:", square)
```

## 5.3 yield

The `yield` keyword is used inside a generator function.
Unlike `return`, `yield` pauses the function and remembers its state,
allowing it to continue from where it stopped.

```python
def even_numbers(limit):
    for number in range(1, limit + 1):
        if number % 2 == 0:
            yield number


numbers = even_numbers(10)

print("Even numbers:")

for number in numbers:
    print(number)
```

## 5.4 Decorators

A decorator is a function that modifies or extends the behavior
of another function without changing its original code.

```python
def log_function(func):
    def wrapper():
        print("Function execution started")
        func()
        print("Function execution completed")
    
    return wrapper


@log_function
def show_message():
    print("Welcome to Python programming")


show_message()
```

```python
def check_access(func):
    def wrapper(role):
        if role == "admin":
            return func(role)
        else:
            print("Access denied")
    
    return wrapper


@check_access
def admin_dashboard(role):
    print("Welcome to the admin dashboard")


admin_dashboard("admin")
admin_dashboard("user")
```

## 5.6 Closures

A closure occurs when an inner function remembers and uses variables
from its outer function even after the outer function has finished execution.

```python
def create_multiplier(multiplier):
    
    def multiply(number):
        return number * multiplier
    
    return multiply


double = create_multiplier(2)
triple = create_multiplier(3)

print("Double of 10:", double(10))
print("Triple of 10:", triple(10))
```

## 5.6 global and nonlocal

The `global` keyword is used to modify a variable defined outside a function
at the global scope.

The `nonlocal` keyword is used inside a nested function to modify a variable
defined in its enclosing function.

```python
# Example of global

counter = 0

def increase_counter():
    global counter
    counter += 1
    return counter


print("Counter:", increase_counter())
print("Counter:", increase_counter())
print("Counter:", increase_counter())
```

## 6.1 map()

The map() function applies a function to every element of an iterable
and returns an iterator containing the results.

```python
numbers = [1, 2, 3, 4, 5]

squared_numbers = list(map(lambda x: x ** 2, numbers))

print("Original numbers:", numbers)
print("Squared numbers:", squared_numbers)
```

## 6.2 filter()

The filter() function selects elements from an iterable based on a condition.

```python
numbers = [10, 15, 20, 25, 30, 35]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Original numbers:", numbers)
print("Even numbers:", even_numbers)
```

## 6.3 zip()

The zip() function combines two or more iterables element by element.
It pairs the values based on their position.

```python
names = ["Dileep", "Rahul", "Anil", "Kiran"]
marks = [85, 92, 78, 88]

student_details = list(zip(names, marks))

print("Student Details:")

for name, mark in student_details:
    print(name, ":", mark)
```

## 6.4 enumerate()

The enumerate() function is used to iterate through a sequence while
keeping track of the index and the value at the same time.

```python
skills = ["Python", "SQL", "Machine Learning", "Deep Learning"]

print("Skills List:")

for index, skill in enumerate(skills, start=1):
    print(index, "-", skill)
```

## 6.5 sorted()

The sorted() function returns a new list containing the elements of an
iterable in sorted order. It can sort values in ascending or descending order.

```python
numbers = [45, 12, 78, 23, 56, 9]

ascending = sorted(numbers)
descending = sorted(numbers, reverse=True)

print("Original list:", numbers)
print("Ascending order:", ascending)
print("Descending order:", descending)
```

## 6.6 any()

The any() function checks whether at least one element in an iterable
is True. It returns True if any condition is satisfied; otherwise,
it returns False.

```python
marks = [45, 38, 72, 41, 35]

passed_students = [mark >= 50 for mark in marks]

print("Marks:", marks)
print("Pass conditions:", passed_students)
print("At least one student passed:", any(passed_students))
```

## 6.7 all()

The all() function checks whether every element in an iterable is True.
It returns True only when all conditions are satisfied.

```python
marks = [65, 72, 81, 90, 76]

passed_students = [mark >= 50 for mark in marks]

print("Marks:", marks)
print("Pass conditions:", passed_students)
print("All students passed:", all(passed_students))
```

## 6.8 isinstance()

The isinstance() function is used to check whether an object belongs
to a particular data type or class.

It returns True if the object matches the specified type; otherwise,
it returns False.

```python
name = "Dileep"
age = 25
marks = [85, 90, 78]
percentage = 82.5

print("Is name a string?", isinstance(name, str))
print("Is age an integer?", isinstance(age, int))
print("Are marks a list?", isinstance(marks, list))
print("Is percentage a float?", isinstance(percentage, float))
print("Is age a string?", isinstance(age, str))
```

## 6.9 id()

The id() function returns the unique identity of an object during
its lifetime.

It can be used to check whether two variables refer to the same object.

```python
numbers = [10, 20, 30]

same_numbers = numbers
different_numbers = [10, 20, 30]

print("ID of numbers:", id(numbers))
print("ID of same_numbers:", id(same_numbers))
print("ID of different_numbers:", id(different_numbers))

print("\nDo numbers and same_numbers have the same ID?",
      id(numbers) == id(same_numbers))

print("Do numbers and different_numbers have the same ID?",
      id(numbers) == id(different_numbers))
```

## 2.3. Constructor (__init__)

The __init__() method is a special method in Python that is automatically
called when an object is created.

It is used to initialize the instance variables of an object.

```python
class Employee:
    def __init__(self, name, role, salary):
        self.name = name
        self.role = role
        self.salary = salary

    def display_details(self):
        print("Employee Name:", self.name)
        print("Role:", self.role)
        print("Salary:", self.salary)


employee1 = Employee("Dileep", "AI/ML Engineer", 50000)
employee2 = Employee("Rahul", "Data Analyst", 45000)

employee1.display_details()
print()
employee2.display_details()
```

## 2.4. Instance Variables

Instance variables are variables that belong to an individual object.
They are usually defined inside the __init__() method using self.

```python
class Employee:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def display(self):
        print("Name:", self.name)
        print("Role:", self.role)


employee1 = Employee("Dileep", "AI/ML Engineer")
employee2 = Employee("Rahul", "Data Analyst")

employee1.display()
employee2.display()
```

## 2.5. Class Variables

A class variable is a variable that is shared by all objects of a class.
It is defined inside the class but outside the instance methods.

```python
class Employee:
    company = "Tech Solutions"

    def __init__(self, name, role):
        self.name = name
        self.role = role

    def display(self):
        print("Name:", self.name)
        print("Role:", self.role)
        print("Company:", Employee.company)


employee1 = Employee("Dileep", "AI/ML Engineer")
employee2 = Employee("Rahul", "Data Analyst")

employee1.display()
print()
employee2.display()
```

## 2.6. Instance Methods

Instance methods are methods that operate on the data of a specific object.
They use self as the first parameter.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_details(self):
        print("Name:", self.name)
        print("Salary:", self.salary)

    def increase_salary(self, amount):
        self.salary += amount


employee = Employee("Dileep", 50000)

employee.display_details()

employee.increase_salary(5000)

print("After Salary Increase:")
employee.display_details()
```

## 2.7. Static Methods

A static method is a method that does not depend on instance or class variables.
It is defined using the @staticmethod decorator.

```python
class Calculator:
    
    @staticmethod
    def add_numbers(a, b):
        return a + b

    @staticmethod
    def multiply_numbers(a, b):
        return a * b


result1 = Calculator.add_numbers(20, 10)
result2 = Calculator.multiply_numbers(20, 10)

print("Addition:", result1)
print("Multiplication:", result2)
```

## 2.8. Class Methods

A class method is a method that works with class-level data.
It uses cls as the first parameter and is defined using @classmethod.

```python
class Employee:
    company = "Tech Solutions"

    def __init__(self, name, role):
        self.name = name
        self.role = role

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

    def display(self):
        print("Name:", self.name)
        print("Role:", self.role)
        print("Company:", self.company)


employee1 = Employee("Dileep", "AI/ML Engineer")
employee2 = Employee("Rahul", "Data Analyst")

Employee.change_company("AI Technologies")

employee1.display()
print()
employee2.display()
```

## 2.9. Properties (@property)

The @property decorator allows a method to be accessed like an attribute.
It is commonly used to control access to object data.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self._salary = salary

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        if value > 0:
            self._salary = value
        else:
            print("Salary must be greater than zero")


employee = Employee("Dileep", 50000)

print("Current Salary:", employee.salary)

employee.salary = 60000

print("Updated Salary:", employee.salary)
```

## 2.10. Encapsulation

Encapsulation is the concept of bundling data and methods together inside a class
and controlling access to the internal data.

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance


account = BankAccount("Dileep", 10000)

account.deposit(5000)

print("Account Holder:", account.account_holder)
print("Balance:", account.get_balance())
```

## 2.11. Abstraction

Abstraction hides the internal implementation details and exposes only
the necessary functionality to the user.

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car starts using an engine")


class ElectricCar(Vehicle):

    def start(self):
        print("Electric car starts using a battery")


car = Car()
electric_car = ElectricCar()

car.start()
electric_car.start()
```

## 2.12. Inheritance

Inheritance allows a child class to inherit attributes and methods
from a parent class.

It promotes code reusability.

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def display_name(self):
        print("Employee Name:", self.name)


class Developer(Employee):
    def write_code(self):
        print(self.name, "is writing Python code")


developer = Developer("Dileep")

developer.display_name()
developer.write_code()
```

## 2.13. Multiple Inheritance

Multiple inheritance allows a child class to inherit
features from two or more parent classes.

```python
class Programmer:
    def write_code(self):
        print("Writing Python code")


class DataScientist:
    def analyze_data(self):
        print("Analyzing data")


class AIEngineer(Programmer, DataScientist):
    def build_model(self):
        print("Building a machine learning model")


engineer = AIEngineer()

engineer.write_code()
engineer.analyze_data()
engineer.build_model()
```

## 2.14. Multilevel Inheritance

Multilevel inheritance occurs when a class inherits from another derived class,
forming a chain of inheritance.

```python
class Person:
    def show_person(self):
        print("This is a person")


class Employee(Person):
    def show_employee(self):
        print("This is an employee")


class Developer(Employee):
    def show_developer(self):
        print("This is a developer")


developer = Developer()

developer.show_person()
developer.show_employee()
developer.show_developer()
```

## 2.15. Hierarchical Inheritance

Hierarchical inheritance occurs when multiple child classes
inherit from the same parent class.

```python
class Employee:
    def show_company(self):
        print("Company: Tech Solutions")


class Developer(Employee):
    def write_code(self):
        print("Developer writes Python code")


class Tester(Employee):
    def test_application(self):
        print("Tester tests the application")


developer = Developer()
tester = Tester()

developer.show_company()
developer.write_code()

print()

tester.show_company()
tester.test_application()
```

## 2.16. Polymorphism

Polymorphism means "many forms".

It allows the same method or interface to perform different
actions depending on the object using it.

```python
class Developer:
    def work(self):
        print("Developer writes and maintains code")


class DataScientist:
    def work(self):
        print("Data Scientist analyzes data and builds models")


class Tester:
    def work(self):
        print("Tester tests the application")


employees = [Developer(), DataScientist(), Tester()]

for employee in employees:
    employee.work()
```

## 2.17. Method Overriding

Method overriding occurs when a child class provides its own
implementation of a method that is already defined in the parent class.

```python
class Employee:
    def work(self):
        print("Employee performs general work")


class Developer(Employee):
    def work(self):
        print("Developer writes Python code")


class Manager(Employee):
    def work(self):
        print("Manager manages the development team")


employee = Employee()
developer = Developer()
manager = Manager()

employee.work()
developer.work()
manager.work()
```

## 2.18. Method Overloading (Python Approach)

Python does not support traditional method overloading.
We can achieve similar behavior using default arguments or *args.

```python
class Calculator:

    def add(self, a, b=0, c=0):
        return a + b + c


calculator = Calculator()

print("Two numbers:", calculator.add(10, 20))
print("Three numbers:", calculator.add(10, 20, 30))
```

## 2.19. Composition

Composition is a "has-a" relationship where one class contains
an object of another class.

The contained object is strongly associated with the main object.

```python
class Engine:
    def start(self):
        print("Engine started")


class Car:
    def __init__(self):
        self.engine = Engine()

    def start_car(self):
        self.engine.start()
        print("Car started")


car = Car()
car.start_car()
```

## 2.20. Aggregation

Aggregation is a weak "has-a" relationship between classes.

The contained objects can exist independently of the object that uses them.

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Employee:", self.name)


class Department:
    def __init__(self, name, employees):
        self.name = name
        self.employees = employees

    def display_employees(self):
        print("Department:", self.name)
        for employee in self.employees:
            employee.display()


employee1 = Employee("Dileep")
employee2 = Employee("Rahul")

employees = [employee1, employee2]

department = Department("AI/ML", employees)
department.display_employees()
```

## 2.21. Association

Association represents a relationship between two independent classes.

The objects can exist independently but interact with each other.

```python
class Student:
    def __init__(self, name):
        self.name = name


class Teacher:
    def __init__(self, name):
        self.name = name

    def teach(self, student):
        print(self.name, "is teaching", student.name)


student = Student("Dileep")
teacher = Teacher("Anil")

teacher.teach(student)
```

## 2.22. Magic / Dunder Methods

Magic methods, also called dunder methods, are special methods
with double underscores before and after their names.

They allow objects to work with Python's built-in operations.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"Student: {self.name}, Marks: {self.marks}"

    def __len__(self):
        return len(self.marks)


student = Student("Dileep", [85, 90, 88])

print(student)
print("Number of subjects:", len(student))
```

## 2.23. super()

The super() function is used to access methods and attributes
of the parent class from a child class.

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Employee Name:", self.name)


class Developer(Employee):
    def __init__(self, name, language):
        super().__init__(name)
        self.language = language

    def display(self):
        super().display()
        print("Programming Language:", self.language)


developer = Developer("Dileep", "Python")
developer.display()
```

## 2.24. __str__()

The __str__() method defines a user-friendly string representation
of an object.

It is automatically called when the object is passed to print().

```python
class Employee:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def __str__(self):
        return f"{self.name} - {self.role}"


employee = Employee("Dileep", "AI/ML Engineer")

print(employee)
```

## 2.25. __repr__()

The __repr__() method provides a detailed representation of an object,
mainly intended for developers and debugging.

```python
class Employee:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def __repr__(self):
        return f"Employee(name='{self.name}', role='{self.role}')"


employee = Employee("Dileep", "AI/ML Engineer")

print(repr(employee))
```

## 2.26. __len__()

The __len__() method allows an object to define its own length.

It is automatically called when the len() function is used.

```python
class Student:
    def __init__(self, subjects):
        self.subjects = subjects

    def __len__(self):
        return len(self.subjects)


student = Student(["Python", "SQL", "Machine Learning", "Deep Learning"])

print("Number of Subjects:", len(student))
```

## 2.27. Dataclasses

Dataclasses help reduce boilerplate code by automatically generating
methods such as __init__(), __repr__() and __eq__().

```python
from dataclasses import dataclass

@dataclass
class Employee:
    name: str
    role: str
    salary: int


employee = Employee(
    name="Dileep",
    role="AI/ML Engineer",
    salary=50000
)

print(employee)
```

## 2.28. Singleton Pattern

The Singleton Pattern ensures that only one instance of a class
is created throughout the program.

It is useful when we need a single shared object, such as a
configuration manager or application settings.

```python
class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance


object1 = Singleton()
object2 = Singleton()

print("Object 1:", object1)
print("Object 2:", object2)
print("Same Object:", object1 is object2)
```

## 3.1 try

The try block contains code that may raise an exception.
Python attempts to execute the statements inside the try block.

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print("Result:", result)
except ValueError:
    print("Please enter a valid number.")
except ZeroDivisionError:
    print("Number cannot be zero.")
```

## 3.2 except

The except block is used to catch and handle exceptions
that occur inside the try block.

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print("Result:", result)

except ValueError:
    print("Invalid input. Please enter a number.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

## 3.3 else

The else block executes when the try block completes successfully
without raising an exception.

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number

except ValueError:
    print("Invalid input.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Calculation successful.")
    print("Result:", result)
```

## 3.4 finally

The finally block always executes whether an exception occurs or not.
It is commonly used for cleanup operations.

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print("Result:", result)

except ValueError:
    print("Invalid input.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

finally:
    print("Execution completed.")
```

## 3.5 raise

The raise statement is used to manually raise an exception
when a specific condition is not satisfied.

```python
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or above.")

    print("Eligible for registration.")


try:
    check_age(20)
except ValueError as error:
    print("Error:", error)
```

## 3.6 assert

The assert statement checks whether a condition is true.
If the condition is false, Python raises an AssertionError.

```python
def calculate_discount(price):
    assert price > 0, "Price must be greater than zero."

    discount = price * 0.10
    return price - discount


final_price = calculate_discount(1000)

print("Final Price:", final_price)
```

## 3.7 Custom Exceptions

Custom exceptions are user-defined exceptions created for
specific application requirements.

They are created by inheriting from the built-in Exception class.

```python
class InsufficientBalanceError(Exception):
    pass


class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient account balance.")

        self.balance -= amount
        print("Withdrawal successful.")
        print("Remaining Balance:", self.balance)


account = BankAccount(10000)

try:
    account.withdraw(12000)
except InsufficientBalanceError as error:
    print("Error:", error)
```

## 3.8 ValueError

ValueError occurs when a function receives a value with the correct
data type but an inappropriate value.

Example: converting a non-numeric string into an integer.

```python
try:
    age = int("twenty")
    print("Age:", age)

except ValueError as error:
    print("ValueError handled:", error)
```

KeyError occurs when we try to access a dictionary using a key

AttributeError occurs when we try to access an attribute or method

# Module 4 – Logging

## 4.1 What is Logging?

Logging is the process of recording information about the execution
of a program.

Python provides the built-in logging module to record messages,
warnings, errors, and other useful information.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Application started")
logging.warning("This is a warning message")
logging.error("An error occurred")
```

## 4.2 Why Use Logging?

Logging is used to monitor application execution, track important events,
identify errors, and troubleshoot problems.

It is more flexible and manageable than using print() statements.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s: %(message)s"
)

user = "Dileep"

logging.info("User login started")
logging.info(f"User {user} logged in successfully")
logging.warning("Password will expire soon")
```

## 4.3 print() vs Logging

print() is mainly used to display simple messages during development.

Logging is more suitable for real applications because it supports
different log levels, formatting, timestamps, and multiple destinations.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

print("Application started")

logging.info("Application started")
logging.warning("Low memory detected")
logging.error("Unable to process the request")
```

## 4.4 Log Levels

Log levels indicate the severity or importance of a log message.

The standard logging levels are:

- DEBUG
- INFO
- WARNING
- ERROR
- CRITICAL

```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(levelname)s - %(message)s"
)

logging.debug("Debugging application flow")
logging.info("Application started successfully")
logging.warning("Memory usage is high")
logging.error("Failed to process the request")
logging.critical("System failure detected")
```

## 4.5 Log Formatters

A log formatter defines the format and information included
in a log message, such as timestamp, log level, and message.

```python
import logging

logger = logging.getLogger("Application")
logger.setLevel(logging.INFO)

handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(asctime)s - %(name)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

logger.info("Application started")
logger.warning("Memory usage is high")
```

## 4.6 File Handlers

A FileHandler is used to write logging messages to a file.

It is useful for maintaining a persistent record of application
events, warnings, and errors.

```python
import logging

logger = logging.getLogger("FileLogger")
logger.setLevel(logging.INFO)

file_handler = logging.FileHandler("application.log")

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

file_handler.setFormatter(formatter)
logger.addHandler(file_handler)

logger.info("Application started")
logger.warning("Low memory warning")
logger.error("An error occurred")

print("Logs written to application.log")
```

## 4.7 Console Handlers

A ConsoleHandler is used to send logging messages to the console.

It is useful for monitoring application activity while the program
is running.

```python
import logging

logger = logging.getLogger("ConsoleLogger")
logger.setLevel(logging.INFO)

console_handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)
logger.addHandler(console_handler)

logger.info("Application started")
logger.warning("Memory usage is high")
logger.error("An error occurred")
```

## 4.8 Rotating Logs

Rotating logs prevent log files from growing indefinitely.

RotatingFileHandler creates backup log files when the current
log file reaches a specified size.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("RotatingLogger")
logger.setLevel(logging.INFO)

# Remove existing handlers to avoid file-locking and duplicate logs
for handler in logger.handlers[:]:
    handler.close()
    logger.removeHandler(handler)

handler = RotatingFileHandler(
    "rotating_application.log",
    maxBytes=200,
    backupCount=3
)

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

for i in range(20):
    logger.info(f"Application event number {i}")

handler.close()

print("Rotating logs created successfully.")
```

Good logging practices make applications easier to monitor,
debug, and maintain.

Important practices include:

- Use appropriate log levels.
- Write meaningful log messages.
- Include useful contextual information.
- Never log passwords, API keys, or sensitive information.
- Use logging instead of excessive print() statements.
- Use file or console handlers as required.
- Use log rotation to control log file size.
